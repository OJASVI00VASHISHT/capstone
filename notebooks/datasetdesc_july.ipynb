```python
import os
import json
from collections import Counter
from PIL import Image
import matplotlib.pyplot as plt

# =========================
# Paths
# =========================
JSON_PATH = "july/instances_default.json"
IMAGE_DIR = "july-images"

# =========================
# Load COCO JSON
# =========================
with open(JSON_PATH, "r") as f:
    coco = json.load(f)

images = coco["images"]
annotations = coco["annotations"]
categories = coco["categories"]

# =========================
# Basic Dataset Info
# =========================
print("=" * 50)
print("DATASET SUMMARY")
print("=" * 50)

print(f"Total Images       : {len(images)}")
print(f"Total Annotations  : {len(annotations)}")
print(f"Total Categories   : {len(categories)}")

# =========================
# Categories
# =========================
print("\nCATEGORIES")
print("-" * 50)

category_map = {}

for cat in categories:
    category_map[cat["id"]] = cat["name"]
    print(f'ID {cat["id"]} --> {cat["name"]}')

# =========================
# Annotation Distribution
# =========================
print("\nANNOTATIONS PER CLASS")
print("-" * 50)

class_counter = Counter()

for ann in annotations:
    class_counter[ann["category_id"]] += 1

for class_id, count in class_counter.items():
    print(f'{category_map[class_id]} : {count}')

# =========================
# Image Resolution Stats
# =========================
widths = []
heights = []

for img in images:
    widths.append(img["width"])
    heights.append(img["height"])

print("\nIMAGE SIZE STATS")
print("-" * 50)

print(f"Min Width   : {min(widths)}")
print(f"Max Width   : {max(widths)}")
print(f"Avg Width   : {sum(widths)/len(widths):.2f}")

print(f"Min Height  : {min(heights)}")
print(f"Max Height  : {max(heights)}")
print(f"Avg Height  : {sum(heights)/len(heights):.2f}")

# =========================
# Bounding Box Stats
# =========================
bbox_widths = []
bbox_heights = []

for ann in annotations:
    bbox = ann["bbox"]
    bbox_widths.append(bbox[2])
    bbox_heights.append(bbox[3])

print("\nBOUNDING BOX STATS")
print("-" * 50)

print(f"Avg Box Width  : {sum(bbox_widths)/len(bbox_widths):.2f}")
print(f"Avg Box Height : {sum(bbox_heights)/len(bbox_heights):.2f}")

# =========================
# Objects Per Image
# =========================
objects_per_image = Counter()

for ann in annotations:
    objects_per_image[ann["image_id"]] += 1

obj_counts = list(objects_per_image.values())

print("\nOBJECTS PER IMAGE")
print("-" * 50)

print(f"Min Objects/Image : {min(obj_counts)}")
print(f"Max Objects/Image : {max(obj_counts)}")
print(f"Avg Objects/Image : {sum(obj_counts)/len(obj_counts):.2f}")

# =========================
# Visualize Sample Images
# =========================
print("\nShowing sample images...")

sample_images = images[:4]

plt.figure(figsize=(12, 12))

for i, img_info in enumerate(sample_images):
    img_path = os.path.join(IMAGE_DIR, img_info["file_name"])

    img = Image.open(img_path)

    plt.subplot(2, 2, i + 1)
    plt.imshow(img)
    plt.title(img_info["file_name"])
    plt.axis("off")

plt.tight_layout()
plt.show()

```
